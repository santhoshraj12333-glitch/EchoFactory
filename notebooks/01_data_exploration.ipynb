```python
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from IPython.display import Audio
```

```python
project_root = Path.cwd().parent

normal_dir = project_root / "data" / "raw" / "pump" / "id_00" / "normal"

wav_files = sorted(normal_dir.glob("*.wav"))

print(f"Found {len(wav_files)} WAV files")

audio_path = wav_files[0]

print("Loading:", audio_path.name)

y, sr = librosa.load(audio_path, sr=16000)

print("Sample Rate:", sr)
print("Duration:", len(y) / sr, "seconds")
print("Samples:", len(y))
```

```python
Audio(y, rate=sr)
```

```python
plt.figure(figsize=(14,4))

librosa.display.waveshow(y, sr=sr)

plt.title("Normal Pump Waveform")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")

plt.show()
```

```python
mel = librosa.feature.melspectrogram(
    y=y,
    sr=sr,
    n_fft=1024,
    hop_length=512,
    n_mels=128
)

mel_db = librosa.power_to_db(mel, ref=np.max)
```

```python
plt.figure(figsize=(14,5))

librosa.display.specshow(
    mel_db,
    sr=sr,
    x_axis="time",
    y_axis="mel"
)

plt.colorbar(format="%+2.0f dB")
plt.title("Mel Spectrogram - Normal Pump")

plt.show()
```
