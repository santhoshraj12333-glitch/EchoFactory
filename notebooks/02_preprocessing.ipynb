```python
import librosa
import numpy as np
from pathlib import Path
from tqdm import tqdm
```

```python
def extract_mel_spectrogram(audio_path,
                            sr=16000,
                            n_mels=128,
                            n_fft=1024,
                            hop_length=512):

    y, sr = librosa.load(audio_path, sr=sr)

    mel = librosa.feature.melspectrogram(
        y=y,
        sr=sr,
        n_fft=n_fft,
        hop_length=hop_length,
        n_mels=n_mels
    )

    mel_db = librosa.power_to_db(mel, ref=np.max)

    return mel_db
```

```python
project_root = Path.cwd().parent

sample = (
    project_root
    / "data"
    / "raw"
    / "pump"
    / "id_00"
    / "normal"
    / "00000000.wav"
)

feature = extract_mel_spectrogram(sample)

print(feature.shape)
```
